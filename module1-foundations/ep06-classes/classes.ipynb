```python
class Sensor:
    def __init__(self, room, temp):
        self.room = room
        self.temp = temp

    def describe(self):
        return f"{self.room}: {self.temp} C"

bedroom = Sensor("bedroom", 21.8)
kitchen = Sensor("kitchen", 24.6)
print(bedroom.describe())
print(kitchen.temp)
```

```python
class Counter:
    def __init__(self):
        self.count = 0

    def add(self, n=1):
        self.count += n
        return self

counter = Counter()
counter.add().add(5)
print(counter.count)
```
