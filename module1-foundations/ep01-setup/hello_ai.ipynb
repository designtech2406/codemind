```python
import pandas as pd

data = pd.DataFrame({
    "hours_studied": [1, 2, 3, 4, 5, 6],
    "exam_score": [52, 58, 65, 70, 78, 85],
})
data
```

```python
import matplotlib.pyplot as plt

plt.scatter(data["hours_studied"], data["exam_score"])
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("More study, higher score?")
plt.show()
```

```python
import numpy as np

slope, intercept = np.polyfit(data["hours_studied"], data["exam_score"], 1)
```

```python
print(slope, intercept)
```

```python
hours = 7
print(f"Predicted score for {hours} hours: {slope * hours + intercept:.1f}")
```
