```python
print(type(172))
print(type(0.86))
print(type("Spam filter v1"))
print(type(True))
```

```python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)    # divide: always a float
print(7 // 2)   # whole-number divide
print(7 % 2)    # remainder
print(7 ** 2)   # power
```

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

```python
import math

print(round(0.1 + 0.2, 2) == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
```

```python
print(int("172") + 1)
print(float("0.86") * 100)
print(str(86) + "%")
print(int(3.9))
```
