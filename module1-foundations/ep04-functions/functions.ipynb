```python
def average(values):
    return sum(values) / len(values)

print(average([52, 58, 65, 70, 78, 85]))
```

```python
def accuracy(correct, total):
    return correct / total

result = accuracy(172, 200)
print(result * 100)
```

```python
def show_accuracy(correct, total):
    print(correct / total)

value = show_accuracy(172, 200)
print(value)
```

```python
def percent(value, digits=1):
    return round(value * 100, digits)

print(percent(0.86427))
print(percent(0.86427, digits=3))
```

```python
def stats(values):
    return min(values), max(values), sum(values) / len(values)

low, high, avg = stats([52, 58, 65, 70, 78, 85])
print(low, high, avg)
```

```python
predictions = [("cat", 0.12), ("dog", 0.81), ("fox", 0.07)]
print(max(predictions, key=lambda p: p[1]))
print(sorted(predictions, key=lambda p: p[1], reverse=True)[:2])
```
