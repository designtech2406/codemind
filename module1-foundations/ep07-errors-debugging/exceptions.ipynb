```python
for text in ["42", "3.5", "", "abc"]:
    try:
        print(float(text))
    except ValueError as e:
        print("skipped:", e)
```

```python
def safe_average(values):
    if not values:
        raise ValueError("no values to average")
    return sum(values) / len(values)

try:
    safe_average([])
except ValueError as e:
    print("Error:", e)
```
