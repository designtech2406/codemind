```python
with open("notes.txt", "a") as f:
    f.write("Model v3: 93% accurate\n")

with open("notes.txt") as f:
    for line in f:
        print(line.strip())
```

```python
import json

settings = {"model": "spam-filter", "learning_rate": 0.01, "epochs": 5, "labels": ["ham", "spam"]}
with open("settings.json", "w") as f:
    json.dump(settings, f, indent=2)

with open("settings.json") as f:
    loaded = json.load(f)
print(loaded["epochs"], loaded["labels"], type(loaded))
```

```python
import csv

rows = [{"hours": 1, "score": 52}, {"hours": 2, "score": 58}, {"hours": 3, "score": 65}]
with open("scores.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["hours", "score"])
    writer.writeheader()
    writer.writerows(rows)

with open("scores.csv") as f:
    for row in csv.DictReader(f):
        print(row)
```
